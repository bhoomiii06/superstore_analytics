```python
import pandas as pd
import numpy as np
import os

print("--- Starting Data Processing Pipeline ---")

# 1. Secure Data Loading
filename = '../data/Superstore Dataset.csv' # Adjust path if running inside 'notebooks' folder

try:
    # Try reading with windows-1252 (common for retail datasets)
    df = pd.read_csv(filename, encoding='windows-1252')
    print(f"SUCCESS: Dataset loaded with windows-1252 encoding. Total Rows: {len(df)}")
except Exception as e:
    # Fallback to standard utf-8
    df = pd.read_csv(filename, encoding='utf-8')
    print(f"SUCCESS: Dataset loaded with utf-8 encoding. Total Rows: {len(df)}")

# 2. Data Cleaning
# Remove duplicates
duplicates = df.duplicated().sum()
df.drop_duplicates(inplace=True)
print(f"Removed {duplicates} duplicate rows.")

# Handle missing Postal Codes
df['Postal Code'] = df['Postal Code'].fillna(0).astype(int)

# 3. Feature Engineering (Crucial for Dashboarding)
# Standardize dates
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Extract time components for time-series analysis
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month_name()
df['Order Quarter'] = "Q" + df['Order Date'].dt.quarter.astype(str)

# Calculate Business Metrics
df['Profit Margin (%)'] = (df['Profit'] / df['Sales']) * 100
df['Shipping Duration (Days)'] = (df['Ship Date'] - df['Order Date']).dt.days

# 4. Export the clean data for Streamlit
output_path = '../data/Final_Analysis.csv'
df.to_csv(output_path, index=False)
print(f"SUCCESS: Cleaned data exported to {output_path} for Streamlit App.")
print("--- Pipeline Complete ---")
```
